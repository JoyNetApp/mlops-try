```python
#mean sales of every category

#3 categories--> Technology, furniture, office supplies
```

```python
#first form a group and then check the average of the group
```

```python
df.groupby(["Category"])['Sales'].mean()
```

```python
#sort it in ascending order

df.groupby(["Category"])['Sales'].mean().sort_values()
```

```python
#sort it in descending order
df.groupby(["Category"])['Sales'].mean().sort_values(ascending=False)
```

```python
#get the category with max average sales

df.groupby(["Category"])['Sales'].mean().sort_values(ascending=True).tail(1)
```

```python

df.groupby(["Category"])['Sales'].mean().sort_values(ascending=False).head(1)
```

```python
#Find the sub-category with the max profit

# At the sub category level I need the total profit

df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False).head(1)
```
